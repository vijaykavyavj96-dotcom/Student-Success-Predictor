```python
import matplotlib.pyplot as plt
```

```python
df20=df.head(20)
```

```python
plt.scatter(df20["Attendance (%)"],df20["FinalGrade"],color="red")
plt.xlabel("AttendanceRate")
plt.ylabel("FinalGrade")
plt.title("Studen Result")
plt.show()
```

```python

df10=df.head(10)
plt.plot(df10["Attendance (%)"],df10["FinalGrade"],"g*",linestyle="dotted",markersize=20)
plt.xlabel("Attendance")
plt.ylabel("Final Grade")
plt.title("Student result analysis")
plt.show()
```

```python
df15=df.head(15)
plt.figure(figsize=(10,5))
plt.bar(df15["StudyHoursPerWeek"],df15["FinalGrade"],color="purple")
plt.xlabel("Study Hours")
plt.ylabel("Final Grade")
plt.title("Study Hours Vs Final Grade")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df['FinalGr-ade'],
         color='skyblue',
         edgecolor='black')

plt.title('Distribution of Final Marks')
plt.xlabel('Final Marks')
plt.ylabel('Number of Students')

plt.show()
```

```python
l=['John','Sarah','Alex','Michael']
FinalGrade=[80.0,87.0,68.0,92.0]
c=['gold','yellowgreen','lightcoral','lightskyblue']
e=(0,0.2,0.4,0.3)
plt.pie(FinalGrade,labels=l,colors=c,explode=e,autopct="%1.2f%%",shadow=True)
plt.title(" Student Final Grade")
plt.show()
```

```python
l=['John','Sarah','Alex','Michael']
PreviousGrade=[78.0,85.0,65.0,90.0]
plt.pie(PreviousGrade,labels=l,autopct="%2.2f%%",shadow=True)
plt.title("Previous Grade Chart")
plt.show()
```
